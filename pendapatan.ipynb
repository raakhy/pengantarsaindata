```python
import pandas as pd

df = pd.read_csv('paket_a_transaksi_pos_2024_2025.csv')

display(df.head())
```

```python
df['kode_kasir'] = df['kode_kasir'].fillna('TIDAK_DIPILIH')
```

```python
df['harga_satuan'] = df['harga_satuan'].astype(str).str.replace(',', '.').astype(float)
```

```python
df['tanggal'] = pd.to_datetime(df['tanggal'], format='mixed', errors='coerce')
df['tahun'] = df['tanggal'].dt.year
```

```python
print("Jumlah Missing Value setelah dibersihkan:")
print(df.isnull().sum())
```

```python
df_selesai = df[df['status'] == 'SELESAI'].copy()
```

```python
df_selesai['pendapatan_bersih'] = (
    df_selesai['jumlah'] *
    df_selesai['harga_satuan'] *
    (1 - df_selesai['diskon_persen'] / 100)
)
```

```python
pendapatan_per_tahun = df_selesai.groupby('tahun')['pendapatan_bersih'].sum().reset_index()
```

```python
print("Total Pendapatan per Tahun (Status: SELESAI):")
print("-" * 45)
for index, row in pendapatan_per_tahun.iterrows():
    print(f"Tahun {int(row['tahun'])} : Rp {row['pendapatan_bersih']:,.2f}")
print("-" * 45)
```

```python
pendapatan_per_tahun.to_csv('ringkasan_pendapatan_posat.csv', index=False)
print("File ringkasan_pendapatan_pos.csv berhasil dibuat!")
```
